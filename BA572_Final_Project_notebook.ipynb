```python
#Initialize a new variable which only stores the outcome question
#as a list.
poster_cols = [col for col in results_df_trimmed.columns
               if 'Please rate how likely you are to click to watch based on the poster'
               in str(col)]

#Extract the rows of those column from our orignal
#dataframe.
poster_ratings = results_df_trimmed[poster_cols]

#Define a function to determine the group
#based on non-null responses in the 8-column chunks.
#This will be used on the poster ratings column
#so the indicies are accurate.
def assign_group(row):
    if row.iloc[0:8].notna().any():
        return 'Control Group'
    elif row.iloc[8:16].notna().any():
        return 'Treatment Arm 1'
    elif row.iloc[16:24].notna().any():
        return 'Treatment Arm 2'
    else:
        return 'Unknown'

#Apply the function to create the new 'Group' column.
results_df_trimmed['Group'] = poster_ratings.apply(
                                assign_group, axis=1)

#Display the group counts to verify.
#We have at least 30 for each group so we fulfilled
#that requirement.
display(results_df_trimmed['Group'].value_counts())
```

```python
#Initialize a dictionary for the starting
#index for each group in our expeirment.
group_start = {
    'Control Group':   0,
    'Treatment Arm 1': 8,
    'Treatment Arm 2': 16,
}


#Create a new varaible, which would show
#the Groups based on their starting index
#(i.e if it is contorl the starts row would have
#0, if it is treatment arm 1, it would have
#8, etc.).
starts = results_df_trimmed['Group'] \
                      .map(group_start)

#For each person, it checks which group they
#are in, determines the starting column to use
#based on the group(i.e. if it is control it
#would start at 0), then adds 4 extra columns
#to that to get the anime rating.
results_df_trimmed['avg_rating_anime'] = [
    poster_ratings.iloc[i, s:s+4].mean() #Based on the person, and which group he is in, start at column index s, and stop 4 columns to the right, and get the mean.
    for i, s in enumerate(starts) #i is the row/person, s is the group that person is in.
]




#For each person, it checks which group they
#are in, determines the starting column to use
#based on the group(i.e. if it is control it
#would start at 0), then adds 4 extra columns
#to that to get the cartoon rating.
results_df_trimmed['avg_rating_cartoon'] = [
    poster_ratings.iloc[i, s+4:s+8].mean()  #Based on the person, and which group he is in, start at column index s, and stop 4 columns to the right, and get the mean.
    for i, s in enumerate(starts) #i is the row/person, s is the group that person is in.
]


```

```python
#Survey data from Qualtrics came in as string(text) and
#not numeric datatype so we have to convert them to numeric.

#List of columns that should function like numbers.
numeric_cols = [
    'Duration', 'Friends_like_anime',
    'Animation_for_kids','Like_animated_shows',
    'Like_anime', 'Like_live_action',
    'Tv_shows_monthly', 'Youtube_usage',
    'Tiktok_instagram_usage','Social_media_active',
    'Location', 'Ethnicity', 'Age_group',
]


#Loop through and apply it to each column
#which should be numeric.
for col in numeric_cols:
    clean_df[col] = pd.to_numeric(clean_df[col],
                                  errors='coerce')
```

```python
#Applying one-hot encoding to change the our genre and
#platform columns, to separte the multiple choices and
#change them into numbers so that our functions can run
#effectively.

#Define a helper function to help with the
#encoding.
def ohe_multiselect(df, col, prefix):

    #Create dummies for the col column,
    #fill its null values with empty string
    #to rpevent crashes, use astype to make
    #sure they are treated as a string,
    #then get dummies with .get_dummeies.
    dummies = (
        df[col]
        .fillna('')
        .astype(str)
        .str.get_dummies(sep=',') #Note we use str.get_dummies because the rows in our columns have stuff like (Tiktok, Snap and not just Tiktok)
    )

    #Remove the empty column we got from fillna().
    dummies = dummies.drop(columns=[''],
                           errors='ignore')  # artifact from NaN rows

    #Rename the columns so that it is clear
    #from which orginal column the dummies
    #came from.
    dummies.columns = [f'{prefix}_{c.strip()}'
                       for c in dummies.columns]

    #Return the dummies.
    return dummies

#Use the helper function to get our dummies.
platform_dummies = ohe_multiselect(clean_df,
                                   'Platform',
                                   'Platform')

genre_dummies = ohe_multiselect(clean_df,
                                   'Genre',
                                   'Genre')

#Add the dummies to our dataframe and remove
#the orignal Platform and Genre columns.
clean_df = pd.concat(
    [clean_df.drop(columns=['Platform', 'Genre']),
     platform_dummies, genre_dummies],
    axis=1,
)
```

##Checking for Randomization

```python
import pingouin as pg
import pandas as pd

# Define the groups
control_group_df = clean_df[clean_df['group'] == 'Control Group']
treatment_arm1_df = clean_df[clean_df['group'] == 'Treatment Arm 1']
treatment_arm2_df = clean_df[clean_df['group'] == 'Treatment Arm 2']

# List of numerical and binary covariates to check for randomization
# Exclude outcome variables and the group column itself
numerical_covariates = [
    'duration', 'friends_like_anime', 'animation_for_kids',
    'like_animated_shows', 'like_anime', 'like_live_action',
    'tv_shows_monthly', 'youtube_usage', 'tiktok_instagram_usage',
    'social_media_active', 'location', 'ethnicity', 'age_group',
    'platform_instagram', 'platform_tiktok', 'platform_youtube',
    'platform_reddit', 'platform_discord', 'platform_facebook',
    'platform_twitter', 'platform_snapchat', 'genre_comedy',
    'genre_action', 'genre_mystery', 'genre_romance', 'genre_sitcom',
    'genre_drama'
]

randomization_check_results = []

print("Performing randomization checks using Pingouin t-test:")

for covariate in numerical_covariates:
    print(f"\n--- Covariate: {covariate} ---")

    # Treatment Arm 1 vs Control Group
    # Use correction=True for Welch's t-test, which does not assume equal variances
    ttest_1_vs_control = pg.ttest(
        x=treatment_arm1_df[covariate],
        y=control_group_df[covariate],
        correction=True
    )
    randomization_check_results.append({
        'Covariate': covariate,
        'Comparison': 'Treatment Arm 1 vs Control Group',
        'p-value': ttest_1_vs_control['p_val'].iloc[0],
        'T': ttest_1_vs_control['T'].iloc[0],
        'df': ttest_1_vs_control['dof'].iloc[0]
    })
    print(f"Treatment Arm 1 vs Control Group:\n{ttest_1_vs_control[['T', 'dof', 'p_val', 'cohen_d']]}")

    # Treatment Arm 2 vs Control Group
    ttest_2_vs_control = pg.ttest(
        x=treatment_arm2_df[covariate],
        y=control_group_df[covariate],
        correction=True
    )
    randomization_check_results.append({
        'Covariate': covariate,
        'Comparison': 'Treatment Arm 2 vs Control Group',
        'p-value': ttest_2_vs_control['p_val'].iloc[0],
        'T': ttest_2_vs_control['T'].iloc[0],
        'df': ttest_2_vs_control['dof'].iloc[0]
    })
    print(f"Treatment Arm 2 vs Control Group:\n{ttest_2_vs_control[['T', 'dof', 'p_val', 'cohen_d']]}")

# Convert results to a DataFrame for better display
randomization_df = pd.DataFrame(randomization_check_results)
print("\n--- Summary of Randomization Check Results ---")
display(randomization_df.round(4))

# Identify covariates with significant differences (e.g., p-value < 0.05)
significant_diffs = randomization_df[randomization_df['p-value'] < 0.05]
if not significant_diffs.empty:
    print("\nCovariates with significant differences (p < 0.05), suggesting potential randomization failure:")
    display(significant_diffs.round(4))
else:
    print("\nNo significant differences found (p < 0.05) between groups for the tested covariates, suggesting successful randomization.")
```

```python
from statsmodels.stats.proportion import proportions_ztest
import pandas as pd
import numpy as np

# Identify numerical covariates (exclude outcome variables)
covariates = [c for c in clean_df.select_dtypes(include=np.number).columns
              if c not in ['avg_rating_anime', 'avg_rating_cartoon']]

# Split data by group
control_df = clean_df[clean_df['Group'] == 'Control Group']
treat1_df = clean_df[clean_df['Group'] == 'Treatment Arm 1']
treat2_df = clean_df[clean_df['Group'] == 'Treatment Arm 2']

results = []

for col in covariates:
    # Check if column is inherently binary (0 and 1 only)
    is_binary = clean_df[col].dropna().isin([0, 1]).all()

    for treat_name, treat_df in [('Treatment Arm 1', treat1_df), ('Treatment Arm 2', treat2_df)]:
        if is_binary:
            # Direct sum for binary fields
            count = np.array([treat_df[col].sum(), control_df[col].sum()])
            nobs = np.array([len(treat_df[col].dropna()), len(control_df[col].dropna())])
        else:
            # For continuous/ordinal, test proportion above the overall median to avoid count > nobs errors
            med = clean_df[col].median()
            count = np.array([(treat_df[col] > med).sum(), (control_df[col] > med).sum()])
            nobs = np.array([len(treat_df[col].dropna()), len(control_df[col].dropna())])

        try:
            stat, pval = proportions_ztest(count, nobs)
        except Exception:
            stat, pval = np.nan, np.nan

        results.append({
            'Covariate': col,
            'Comparison': f'{treat_name} vs Control',
            'Z-Statistic': stat,
            'p-value': pval,
            'Note': 'Binary Test' if is_binary else 'Prop > Median Test'
        })

# Format and display the results neatly
results_df = pd.DataFrame(results)

print("--- Proportions Z-Test Randomization Check ---")
display(results_df.round(4).style.set_properties(**{'text-align': 'left'}).set_table_styles([
    dict(selector='th', props=[('text-align', 'left')])
]))
```

```python
# Filter the results for p-values < 0.05
significant_ztests = results_df[results_df['p-value'] < 0.05]

if significant_ztests.empty:
    print("No z-tests resulted in a p-value less than 0.05.")
else:
    print("Here are the z-tests with a p-value less than 0.05:")
    display(significant_ztests)
```
